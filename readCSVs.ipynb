```python
for i in range(0, 20, 1): 
    df = pd.read_csv(f'C:/BHCC/Internship CSC INT 299/NWS_Data/hooj_request1/HeatRisk-v2.5-USC00023160-{2005+i}.csv')
    # remove duplicates
    df.drop_duplicates(inplace = True)

    # remove all rows with NULL values
    df.dropna(inplace = True)

    # calculate the mean values of the empty cells
    try:
        a = round(df["MaxT_Records"].mean())
        b = round(df["MinT_Records"].mean())
        x = round(df["MaxT_Obs"].mean())
        y = round(df["MinT_Obs"].mean())
        z = round(df["HeatRisk"].mean())

        print(a,b,x,y,z)
    except:
        print('Cannot handle null values.')

# print(df.to_string())
```
